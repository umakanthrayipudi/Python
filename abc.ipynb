```python
from abc import ABC, abstractmethod

# Abstract Expression
class Expression(ABC):
    @abstractmethod
    def interpret(self):
        pass

# Terminal Expression (Number)
class Number(Expression):
    def __init__(self, value):
        self.value = value

    def interpret(self):
        return self.value

# Non-Terminal Expression (Addition)
class Add(Expression):
    def __init__(self, left, right):
        self.left = left
        print(left)
        self.right = right

    def interpret(self):
        return self.left.interpret() + self.right.interpret()

# Non-Terminal Expression (Subtraction)
class Subtract(Expression):
    def __init__(self, left, right):
        self.left = left
        self.right = right

    def interpret(self):
        return self.left.interpret() - self.right.interpret()

# Usage
expr = Subtract(Add(Number(2), Number(3)), Number(1))  # (2 + 3) - 1
print(expr.interpret())  # Output: 4
```

```python
# A decorator to add logging functionality to a class
def log_method_calls(cls):
    class Wrapped(cls):
        def __getattribute__(self, name):
            # Wrap methods to log before calling
            attr = super().__getattribute__(name)
            if callable(attr):
                def wrapper(*args, **kwargs):
                    print(f"Calling method: {name}")
                    return attr(*args, **kwargs)
                return wrapper
            return attr
    return Wrapped

# Using the decorator
@log_method_calls
class Person:
    def __init__(self, name):
        self.name = name
    
    def say_hello(self):
        return f"Hello, my name is {self.name}"

# Client code
person = Person("Alice")
print(person.say_hello())
```

```python
class Handler:
    def __init__(self):
        self.next_handler = None

    def set_next(self, handler):
        self.next_handler = handler
        return handler

    def handle(self, request):
        if self.next_handler:
            self.next_handler.handle(request)


class ConcreteHandlerA(Handler):
    def handle(self, request):
        if request == "A":
            print("Handled by ConcreteHandlerA")
        elif self.next_handler:
            self.next_handler.handle(request)


class ConcreteHandlerB(Handler):
    def handle(self, request):
        if request == "B":
            print("Handled by ConcreteHandlerB")
        elif self.next_handler:
            self.next_handler.handle(request)


# Example Usage
handler1 = ConcreteHandlerA()
handler2 = ConcreteHandlerB()

handler1.set_next(handler2)

handler1.handle("A")  # Output: Handled by ConcreteHandlerA
handler1.handle("B")  # Output: Handled by ConcreteHandlerB
handler1.handle("C")  # No handler found
```

```python
# Command interface
class Command:
    def execute(self):
        pass

# ConcreteCommand
class LightOnCommand(Command):
    def __init__(self, light):
        self.light = light
    
    def execute(self):
        self.light.turn_on()

class LightOffCommand(Command):
    def __init__(self, light):
        self.light = light
    
    def execute(self):
        self.light.turn_off()

# Receiver
class Light:
    def turn_on(self):
        print("The light is ON")

    def turn_off(self):
        print("The light is OFF")

# Invoker
class RemoteControl:
    def __init__(self):
        self.command = None
    
    def set_command(self, command):
        self.command = command
    
    def press_button(self):
        self.command.execute()

# Example Usage
light = Light()

light_on = LightOnCommand(light)
light_off = LightOffCommand(light)

remote = RemoteControl()
remote.set_command(light_on)
remote.press_button()  # Output: The light is ON

remote.set_command(light_off)
remote.press_button()  # Output: The light is OFF
```

```python
class Solution:
    def equalPairs(self, grid: list[list[int]]) -> int:
        length = len(grid)
        rows,colums = self.calculate_sum(grid=grid)
        new_list = 0
        for i in range(0,length):
            for j in range(0,length):
                if rows[i] == colums[j] and self.check_rows_columns(grid=grid,row=i,column=j):
                    new_list+=1
        return new_list        

    def calculate_sum(self, grid: list[list[int]]) :
        length = len(grid)
        rows=[]
        colums=[]
        for i in range(0,length):
            s = 0
            s2 = 0
            for j in range(0,length):
                s+=grid[i][j] # left to rigth
                s2+=grid[j][i] # right to left
            rows.append(s)
            colums.append(s2)
        return rows,colums
    
    def check_rows_columns(self, grid: list[list[int]], row, column) :
        length = len(grid)
        for j in range(0,length):
            if grid[row][j] != grid[j][column]:
                return False
        return True

sol = Solution()
res =sol.equalPairs([[3,2,1],[1,7,6],[2,7,7]])
print(res)
```

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1
abc = [3,2,3,1,3,4,5]
merge_sort(abc)
print(abc)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

abc = [3,2,3,1,3,4,5]
abc = quick_sort(abc)
print(abc)
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
from collections import deque
import Optional
class Solution:
    count = 0
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        self.count = 0
        self.iteratedown(root,targetSum,deque())
        return self.count

    def iteratedown(self,root: Optional[TreeNode], targetSum: int, que: deque):
        if root == None:
            return
        newque = deque()
        s = 0
        while(len(que)):
            a=que.popleft()
            newque.append(a)
            s+=a
        newque.append(root.val)
        s+=root.val
        if s >= targetSum:
            while(len(newque)>0):
                if s == targetSum:
                   self.count+=1
                   newque.popleft()
                   break 
                if s < targetSum:
                    break
                if s > targetSum:
                    s-=newque.popleft()
        self.iteratedown(root.left,targetSum, newque)
        self.iteratedown(root.right,targetSum, newque)
        
```

```python
#S: Single Responsibility Principle (SRP)
#A class should have one, and only one, reason to change.
#A class must have only one responsibulity 
# This principle aims to separate behaviours so that if bugs arise as a result of your change,
# it won’t affect other unrelated behaviours.

class UserManger:
    def user_authentication():
        pass
    
    def user_login():
        pass
    
    def send_email_notificatoin():
        pass


# O — Open-Closed
# Classes should be open for extension, but closed for modification
# This principle aims to extend a Class’s behaviour without changing the existing behaviour of that Class.
# This is to avoid causing bugs wherever the Class is being used.
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass
    
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height
    
    def calculate_area(self):
        return self.width * self.height   
   
#  L — Liskov Substitution

# If S is a subtype of T, then objects of type T in a program may be replaced
# with objects of type S without altering any of the desirable properties of that program.

# This principle aims to enforce consistency -
# - so that the parent Class or its child Class can be used in the same way without any errors.

from abc import ABC, abstractmethod

class vehicle():
    def start_engine(self):
        pass
    
class car(vehicle):
    def start_engine(self):
        pass
    
class bicycle(vehicle):
    def start_engine(self):
        pass

class vehicle():
    def start(self):
        pass
    
class car(vehicle):
    def start(self):
        pass
    
class bicycle(vehicle):
    def start(self):
        pass

#  I — Interface Segregation

# Clients should not be forced to depend on methods that they do not use.   

# This principle aims at splitting a set of actions into smaller sets
# so that a Class executes ONLY the set of actions it requires.


# D — Dependency Inversion

# - High-level modules should not depend on low-level modules. Both should depend on the abstraction.

# - Abstractions should not depend on details. Details should depend on abstractions.
class GmailClient:
    def send_email(self, recipient, subject,body):
        pass
    
class EmailService:
    def __init__(self):
        self.gmail_client = GmailClient()
        
    def send_email(self, recipient, subject,body):
        self.gmail_client.send_email(recipient,subject,body)
#-----------------------------------------------------------------
class EmailClient:
    def send_email(self, recipient, subject,body):
        pass

class GmailClient(EmailClient):
    def send_email(self, recipient, subject,body):
        pass
    
class EmailService:
    def __init__(self,email_client):
        self.email_client = email_client
        
    def send_email(self, recipient, subject,body):
        self.email_client.send_email(recipient,subject,body)
        
gmail =  GmailClient()
email = EmailService(gmail)
email.send_email("","","")
```
